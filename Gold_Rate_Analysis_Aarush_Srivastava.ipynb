```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Create / Load the dataset\nFor this beginner version, we create sample monthly values. Gold rate is expressed in INR per 10 grams.

```python
# Create sample monthly data (illustrative only)
dates = pd.date_range(start="2020-01-01", end="2025-12-01", freq="MS")
np.random.seed(42)

rates = []
price = 39000
for date in dates:
    change = 0.0085 + np.random.normal(0, 0.012)  # trend + random variation
    price = price * (1 + change)
    seasonal = 1 + 0.015 * np.sin(2 * np.pi * date.month / 12)
    rates.append(round(price * seasonal, 2))

gold = pd.DataFrame({
    "Date": dates,
    "Gold_Rate_INR_per_10g": rates
})
gold.head()
```

## 2. Inspect the data\nUse `head()`, `tail()`, `info()` and `describe()` to get a quick overview.

```python
print("First 5 rows:")
display(gold.head())

print("Last 5 rows:")
display(gold.tail())

print("Dataset information:")
gold.info()

print("Summary statistics:")
display(gold["Gold_Rate_INR_per_10g"].describe())
```

## 3. Calculate changes and moving averages

```python
gold["Monthly_Change"] = gold["Gold_Rate_INR_per_10g"].diff()
gold["Monthly_Change_Percent"] = gold["Gold_Rate_INR_per_10g"].pct_change() * 100
gold["MA_3_Month"] = gold["Gold_Rate_INR_per_10g"].rolling(window=3).mean()
gold["MA_6_Month"] = gold["Gold_Rate_INR_per_10g"].rolling(window=6).mean()

display(gold.head(10))
```

## 4. Key statistics using NumPy

```python
values = gold["Gold_Rate_INR_per_10g"].to_numpy()

print("Average gold rate:", round(np.mean(values), 2))
print("Median gold rate:", round(np.median(values), 2))
print("Minimum gold rate:", round(np.min(values), 2))
print("Maximum gold rate:", round(np.max(values), 2))
print("Standard deviation:", round(np.std(values), 2))
```

## 5. Year-wise average rate

```python
gold["Year"] = gold["Date"].dt.year
yearly_average = gold.groupby("Year")["Gold_Rate_INR_per_10g"].mean().round(2)
display(yearly_average)
```

## 6. Plot the gold-rate trend

```python
plt.figure(figsize=(12, 5))
plt.plot(gold["Date"], gold["Gold_Rate_INR_per_10g"], marker=".", label="Monthly gold rate")
plt.title("Illustrative Gold Rate Trend (INR per 10g)")
plt.xlabel("Date")
plt.ylabel("Gold rate (INR per 10 grams)")
plt.grid(True)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 7. Compare gold rate with moving averages

```python
plt.figure(figsize=(12, 5))
plt.plot(gold["Date"], gold["Gold_Rate_INR_per_10g"], label="Gold rate")
plt.plot(gold["Date"], gold["MA_3_Month"], label="3-month moving average")
plt.plot(gold["Date"], gold["MA_6_Month"], label="6-month moving average")
plt.title("Gold Rate and Moving Averages")
plt.xlabel("Date")
plt.ylabel("Gold rate (INR per 10 grams)")
plt.grid(True)
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 8. Monthly percentage change

```python
plt.figure(figsize=(12, 4))
plt.bar(gold["Date"], gold["Monthly_Change_Percent"])
plt.title("Month-to-month Gold Rate Percentage Change")
plt.xlabel("Date")
plt.ylabel("Change (%)")
plt.grid(axis="y")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 9. Find the highest and lowest price rows

```python
highest = gold.loc[gold["Gold_Rate_INR_per_10g"].idxmax()]
lowest = gold.loc[gold["Gold_Rate_INR_per_10g"].idxmin()]

print("Highest sample rate:")
display(highest)
print("Lowest sample rate:")
display(lowest)
```
